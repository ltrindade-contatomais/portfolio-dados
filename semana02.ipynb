## Exercicio 4
Dado um valor_venda e uma meta = 1000, use um condicional para imprimir se bateu a meta ou não, e por quantos reais faltou (ou passou).

Por exemplo, com valor_venda = 850, a mensagem deveria dizer algo como "Não bateu a meta, faltam R$ 150" — e com valor_venda = 1200, algo como "Bateu a meta, passou R$ 200".

```python
valor_venda = 1200
meta = 1000

if valor_venda >= meta:
    print(f"Bateu a meta {valor_venda - meta} acima")
else:
    print(f"Não bateu a meta faltando {meta - valor_venda}")
```
