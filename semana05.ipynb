```python

def calcular_comissao(vendas):
    try:
        return vendas * 150
    except:
        return "Erro: número de vendas inválido"

def calcular_comissao(vendas):
    if type(vendas) == int or type(vendas) == float:
        return vendas * 150
    else:
        return "Erro: número de vendas inválido"

    
print(calcular_comissao(5))
print(calcular_comissao("cinco"))
 
```
