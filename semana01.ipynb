25% foi a taxa de ligações atendidas, sabendo qua a quantidade total de ligaçõa foi 2400, precisamos fazer (ligações * 0.25) nesse calculo acharemos a quantidade de ligações atendidas

```python
ligacoes = 2400
taxa = 0.25
atendidas = ligacoes * taxa
print(atendidas)
```

Sabendo que a 600 foram as ligações atendidas e que desse numero 45 foram reuniões, precisamos fazer (600/45*100) assim saberemos qual foi a taxa de conversão entre atendidos para reuniões agendadas

```python
atendidas = 600
reunioes = 45
taxa_conversão = reunioes / atendidas * 100
print(f"taxa_conversão: {taxa_conversão: .1f}%")
```

Sabendo que o faturamento de 8000 passou para 9200, queremos saber qual foi a variação entre eles, para isso precisamos fazer (9200 - 8000) / 8000 * 100 assim chegaremos ao percentual de variação do resultado

```python
passado = 8000
atual = 9200
variação = (atual - passado) / passado * 100
print(f"variação: {variação: .1f}%")
```

Sabendo que o valor do produto é R$400,00 precisamos saber quanto fica com 15% de desconto e 15% de acrescimo no valor, para o desconto a conta é 85% do valor do produto (400 * 0.85) e para o acrescimo a conta é (400 * 1.15)

```python
preçodesc = 400 * 0.85
preçoacresc = 400 * 1.15
print(round(preçodesc, 2))
print(round(preçoacresc, 2))
```

para achar o lucro precisamos fazer vendas - custo e para achar a porcentagem de lucro e vendas precisa dividir o lucro pelo custo *100 e a mesma coisa com as vendas 

```python
custo = 180
vendas = 270
lucro = vendas - custo
lucro_porcentagem = lucro / custo  * 100
vendas_porcentagem = lucro / vendas * 100
print(lucro)
print(f"lucro_porcentagem: {lucro_porcentagem: .1f}%")
print(f"vendas_porcentagem: {vendas_porcentagem: .1f}%")
```
