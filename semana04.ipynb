```python
leads = [
    {"nome": "Farooeste", "origem": "indicação", "taxa_conversao": 12},
     {"nome": "M3s", "origem": "instagram", "taxa_conversao": 9},
     {"nome": "JQL", "origem": "indicação", "taxa_conversao": 6},
     {"nome": "órigo", "origem": "google maps", "taxa_conversao": 4},
     {"nome": "mont car", "origem": "indicação", "taxa_conversao": 15},
    
]

for lead in leads:
    if lead["taxa_conversao"] > 8 and lead["origem"] == "indicação":
        print(f"lead {lead['nome']} prioritário")
    elif lead["taxa_conversao"] >= 5:
        print(f"lead {lead['nome']} acompanhar")
    else:
        print(f"lead {lead['nome']} descartar")
```
